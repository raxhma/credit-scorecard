# Credit Scorecard Tutorial: "Give Me Some Credit"

**Goal:** build a simple, readable *application-style* credit scorecard that estimates the
**Probability of Default (PD)** of a borrower and converts it into a **score** (like a FICO score).

**Target:** `SeriousDlqin2yrs` = 1 if the person was 90+ days past due within the next 2 years (a "bad"), 0 otherwise (a "good").

| Step | What we do |
|---|---|
| 1 | Load the data and look at the default rate |
| 2 | Clean it: missing values and outliers |
| 3 | Stratified train/test split (no leakage) |
| 4 | Bin variables, compute **WoE** and **IV** |
| 5 | Replace raw values by WoE values |
| 6 | Logistic regression on the WoE variables |
| 7 | Convert the model into a **points-based scorecard** (base score 600, PDO 20, odds 50:1) |
| 8 | Evaluate on the test set: AUC, Gini, KS, score distribution, calibration |
| 9 | Save the scorecard for the Streamlit app |

**Mini glossary**
- **PD**: probability that the borrower defaults within the horizon (here 2 years).
- **Good / Bad**: non-defaulter / defaulter. **Odds** = good / bad.
- **WoE** (Weight of Evidence): a number saying how "good" or "bad" a group of borrowers is compared with the average.
- **IV** (Information Value): how useful a variable is overall for separating goods from bads.
- **PDO** (Points to Double the Odds): how many score points make the good:bad odds twice as large.

> **Reading tip:** every section starts with a plain-words explanation, then the code, then (when useful) a short interpretation of the output.

## 1. Load the data and check the default rate

**What it does:** imports the libraries, reads the CSV and prints the share of borrowers who defaulted.

**Why a bank does it:** the *default rate* (also called bad rate) is the first number any credit risk
team looks at. It tells you how rare the event is (here, rare = **imbalanced** data) and gives a baseline:
a model is only useful if it does better than "everybody has the average PD".

```python
import json
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, roc_curve

pd.set_option("display.width", 170)
pd.set_option("display.max_columns", 30)

RANDOM_STATE = 42          # fixed seed so results are reproducible
TARGET = "SeriousDlqin2yrs"
os.makedirs("figures", exist_ok=True)
```

```python
# Read the file. "Unnamed: 0" is just a row number added by Kaggle, so we drop it.
df = pd.read_csv("data/cs-training.csv")
df = df.drop(columns="Unnamed: 0")

print("Shape (rows, columns):", df.shape)
df.head()
```

```python
# Default rate = average of the 0/1 target
default_rate = df[TARGET].mean()
print(f"Default rate: {default_rate:.2%}")
print(df[TARGET].value_counts().rename({0: "good (0)", 1: "bad (1)"}))
```

## 2. Missing values and outliers

**What it does:** counts missing values, looks at extreme values, then applies a few simple, explicit cleaning rules.

**Why a bank does it:** raw data always contains errors, special codes and system defaults. A model trained on
them would learn noise (or worse, nonsense like "age 0 is a good customer"). Banks also need every rule to be
*documented and reproducible* because model validation teams and regulators (IRB, IFRS 9 model governance)
will ask "what exactly did you do to the data?".

```python
# 2a. Missing values: how many, and what share?
missing = df.isna().sum().to_frame("n_missing")
missing["pct_missing"] = (missing["n_missing"] / len(df) * 100).round(2)
missing[missing["n_missing"] > 0]
```

```python
# 2b. Look at the distribution: min, quantiles and max reveal the outliers
df.describe(percentiles=[0.01, 0.5, 0.99]).T[["min", "1%", "50%", "99%", "max"]]
```

**What we see (and what we decide):**

| Problem | Evidence | Rule we apply |
|---|---|---|
| `age` = 0 | impossible value (and a borrower must be an adult) | drop rows with age < 18 |
| Past-due counts equal to 96 or 98 | these look like **special codes**, not real counts (nobody is 98 times late) | cap the counts at 10 |
| `RevolvingUtilizationOfUnsecuredLines` up to 50,000 | utilization is a ratio, normally between 0 and about 1 | cap at 2 (200%) |
| `DebtRatio` up to 329,664 | not a real ratio (see check below) | values above 5 are set to *missing* |
| `MonthlyIncome` missing (about 20%) | income was simply not recorded | keep as missing, give it its own bin later |
| `NumberOfDependents` missing (about 2.6%) | not recorded | fill with the **train** median (done after the split) |

Let's verify the two claims about special codes and `DebtRatio` before we apply the rules.

```python
# Check 1: borrowers with the special code 96/98 in "90 days late"
special = df["NumberOfTimes90DaysLate"] >= 96
print("Rows with the 96/98 code:", special.sum())
print(f"Their default rate: {df.loc[special, TARGET].mean():.1%}  (overall: {default_rate:.1%})")

# Check 2: where does the extreme DebtRatio come from?
huge_ratio = df["DebtRatio"] > 5
print("\nRows with DebtRatio > 5:", huge_ratio.sum())
print(f"Of these, income is missing for {df.loc[huge_ratio, 'MonthlyIncome'].isna().mean():.0%}")
print(f"Default rate of these rows: {df.loc[huge_ratio, TARGET].mean():.1%}")
```

**Interpretation:** the 96/98 code rows are *very* risky, so capping them at 10 (instead of deleting them)
keeps that useful signal: they land in the "6 or more" group. For `DebtRatio`, almost every huge value comes from a
borrower with **missing (or zero) income**: the dataset seems to store the raw debt amount instead of a ratio there.
So the number is not a real ratio and we treat it as *unknown*.

**About data leakage:** the rules below use **fixed thresholds** chosen from business logic, they do not learn anything
from the data (no mean, median, quantile). So it is safe to apply them before the split.
Anything that *learns* from the data (median, bins, WoE, model) will be done **after** the split, on the training set only.

```python
n_before = len(df)

# Rule 1: remove impossible ages
df = df[df["age"] >= 18].copy()

# Rule 2: the three "number of times past due" columns: cap at 10 (handles the 96/98 codes)
past_due_cols = [
    "NumberOfTime30-59DaysPastDueNotWorse",
    "NumberOfTimes90DaysLate",
    "NumberOfTime60-89DaysPastDueNotWorse",
]
for col in past_due_cols:
    df[col] = df[col].clip(upper=10)

# Rule 3: cap the credit utilization ratio at 2 (200%)
df["RevolvingUtilizationOfUnsecuredLines"] = df["RevolvingUtilizationOfUnsecuredLines"].clip(upper=2)

# Rule 4: a debt ratio above 5 is not a real ratio -> set to missing (NaN)
df.loc[df["DebtRatio"] > 5, "DebtRatio"] = np.nan

print("Rows removed:", n_before - len(df))
print("DebtRatio now missing:", df["DebtRatio"].isna().sum())
print(f"Default rate after cleaning: {df[TARGET].mean():.2%}")
```

## 3. Stratified train/test split (no data leakage)

**What it does:** splits the data into 70% **train** (used to build the model) and 30% **test** (kept aside to judge it).
*Stratified* means both parts keep the same default rate. Then we fill the missing `NumberOfDependents`
using the median **of the training set only**.

**Why a bank does it:** the test set simulates "new customers we have never seen". If any information from the
test set leaks into the build (even something small like a median), the performance you measure is too optimistic,
and the model will disappoint in production. Stratification matters because defaults are rare: a random split could
by chance put too few defaults in one part.

```python
X = df.drop(columns=TARGET)
y = df[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    stratify=y,                 # keep the same default rate in both parts
    random_state=RANDOM_STATE,
)

print("Train:", X_train.shape, f"default rate = {y_train.mean():.2%}")
print("Test :", X_test.shape,  f"default rate = {y_test.mean():.2%}")
```

```python
# Fill missing dependents with the TRAIN median, and reuse that same number on the test set.
dependents_median = X_train["NumberOfDependents"].median()   # learned on train only
X_train["NumberOfDependents"] = X_train["NumberOfDependents"].fillna(dependents_median)
X_test["NumberOfDependents"] = X_test["NumberOfDependents"].fillna(dependents_median)

print("Median used:", dependents_median)
print("Remaining missing values in train:")
print(X_train.isna().sum()[lambda s: s > 0])
```

`MonthlyIncome` and `DebtRatio` still contain missing values on purpose: in the next step, "missing" becomes its **own bin**.
This is common in scorecards, because *not having a value* can itself be informative (for example, here a borrower with no
recorded income does not look like the average borrower).

## 4. Binning, Weight of Evidence (WoE) and Information Value (IV)

**What it does:** for each variable we cut its values into a few **bins** (for example age: ≤30, 30-40, ...), then for each bin we compute:

- `dist_good` = share of all goods that fall in the bin, `dist_bad` = share of all bads that fall in the bin
- **WoE** = ln(`dist_good` / `dist_bad`)
  - WoE > 0 → the bin has relatively more goods than average (safer)
  - WoE < 0 → relatively more bads (riskier)
  - WoE = 0 → same as the average
- **IV** = sum over bins of (`dist_good` − `dist_bad`) × WoE → one number per variable telling how well it separates goods from bads.

| IV | Predictive power |
|---|---|
| < 0.02 | useless |
| 0.02 – 0.1 | weak |
| 0.1 – 0.3 | medium |
| 0.3 – 0.5 | strong |
| > 0.5 | suspiciously strong (check for leakage) |

**Why a bank does it:**
- Binning makes the model **robust to outliers** and handles **non-linear** effects (e.g. very young and very old borrowers can both be riskier).
- WoE puts every variable on the **same scale** and is directly linked to the log-odds, so logistic regression on WoE gives a clean and **explainable** scorecard.
- Missing values get their own bin, so no imputation guesses are needed.
- IV is the classic tool for **variable selection**.

**Important:** bins and WoE are computed on the **training set only**, then simply *applied* to the test set.

```python
# Bin edges per variable, chosen by looking at the default rate per bin on the TRAIN set.
# A bin is (lower, upper]. The first bin is "<= first edge", the last one is "> last edge".
bin_edges = {
    "RevolvingUtilizationOfUnsecuredLines": [0.05, 0.15, 0.3, 0.5, 0.7, 0.9, 1.0],
    "age": [30, 40, 50, 60, 70],
    "DebtRatio": [0.4, 0.5, 0.7, 1.0],
    "MonthlyIncome": [3000, 5000, 6500, 8000, 10000],
    "NumberOfOpenCreditLinesAndLoans": [1, 3, 5, 13],
    "NumberRealEstateLoansOrLines": [0, 2],
    "NumberOfDependents": [0, 1, 2],
    "NumberOfTime30-59DaysPastDueNotWorse": [0, 1, 2, 5],
    "NumberOfTimes90DaysLate": [0, 1, 2],
    "NumberOfTime60-89DaysPastDueNotWorse": [0, 1, 2],
}
```

```python
def make_labels(edges):
    # Turns [30, 40, 50] into ["<= 30", "(30, 40]", "(40, 50]", "> 50"]
    labels = [f"<= {edges[0]:g}"]
    for low, high in zip(edges[:-1], edges[1:]):
        labels.append(f"({low:g}, {high:g}]")
    labels.append(f"> {edges[-1]:g}")
    return labels


def assign_bins(series, edges):
    # Put each value in its bin. Missing values go to a bin called "Missing".
    cut = pd.cut(series, bins=[-np.inf] + edges + [np.inf], labels=make_labels(edges))
    return cut.astype(object).fillna("Missing")


def woe_table(bins, y, edges):
    # Build the WoE / IV table of ONE variable.
    t = pd.DataFrame({"bin": bins.values, "is_bad": y.values})
    t = t.groupby("bin")["is_bad"].agg(total="count", bad="sum")
    t["good"] = t["total"] - t["bad"]

    # keep the bins in their natural order (and keep only those that exist)
    order = [b for b in make_labels(edges) + ["Missing"] if b in t.index]
    t = t.loc[order]

    t["pct_of_total"] = t["total"] / t["total"].sum()
    t["default_rate"] = t["bad"] / t["total"]

    # +0.5 avoids log(0) if a bin has no good or no bad (rare, but safe)
    good_smooth = t["good"] + 0.5
    bad_smooth = t["bad"] + 0.5
    t["dist_good"] = good_smooth / good_smooth.sum()
    t["dist_bad"] = bad_smooth / bad_smooth.sum()

    t["woe"] = np.log(t["dist_good"] / t["dist_bad"])
    t["iv_part"] = (t["dist_good"] - t["dist_bad"]) * t["woe"]
    return t
```

```python
woe_tables = {}   # one table per variable
iv_values = {}    # one IV per variable

for var, edges in bin_edges.items():
    bins_train = assign_bins(X_train[var], edges)
    table = woe_table(bins_train, y_train, edges)
    woe_tables[var] = table
    iv_values[var] = table["iv_part"].sum()

iv_summary = pd.Series(iv_values, name="IV").sort_values(ascending=False).to_frame()
iv_summary["power"] = pd.cut(
    iv_summary["IV"], [-np.inf, 0.02, 0.1, 0.3, 0.5, np.inf],
    labels=["useless", "weak", "medium", "strong", "suspicious"],
)
iv_summary.round(4)
```

```python
# Bar chart of the IVs
ax = iv_summary["IV"].sort_values().plot(kind="barh", figsize=(8, 4), color="steelblue")
ax.axvline(0.02, color="red", linestyle="--", label="0.02 threshold")
ax.set_xlabel("Information Value (train set)")
ax.set_title("Which variables separate goods from bads?")
ax.legend()
plt.tight_layout()
plt.savefig("figures/iv_ranking.png", dpi=120)
plt.show()
```

```python
# The detailed WoE table of every variable (sorted by IV)
show_cols = ["total", "pct_of_total", "bad", "default_rate", "woe", "iv_part"]
for var in iv_summary.index:
    print(f"\n=== {var}   (IV = {iv_values[var]:.4f}) ===")
    print(woe_tables[var][show_cols].round(4).to_string())
```

**How to read one table** (take the first one): `default_rate` rises from the safest bin to the riskiest bin and `woe`
goes the opposite way (positive for safe bins, negative for risky bins). A **monotonic** pattern like that is what we want,
because it is easy to explain ("the more of their credit lines a person uses, the riskier they are").

**What the IV ranking says (real numbers from this run):**

- **Strongest:** credit utilization (IV ≈ 1.14) and the three past-due counts (90+ days ≈ 0.89, 30-59 days ≈ 0.74, 60-89 days ≈ 0.60).
- **Medium:** `age` (≈ 0.23).
- **Weak:** open credit lines (≈ 0.10), income (≈ 0.08), debt ratio (≈ 0.07), real-estate loans (≈ 0.06), dependents (≈ 0.03).
- All ten variables are above 0.02, so **none is dropped** by our rule.

**Reading the bins:**

- Utilization: the default rate climbs from about 1.8% (utilization ≤ 5%) to about 37% (above 100%).
- 90+ days late: about 4.6% for borrowers who were never late, 33.7% after one time, about 61% after three or more times.
- Age: about 11.3% for ≤ 30 years old down to about 2.4% above 70.
- **Missing income and missing debt ratio** have default rates of about 5.5% and 5.4%, *below* the 6.7% average, so "not recorded" does not mean "risky" in this dataset. That is exactly why a separate "Missing" bin is better than guessing a value.
- Not every variable is monotonic: real-estate loans are riskier with 0 loans (8.3%) and with more than 2 (8.6%) than with 1 to 2 loans (5.4%).

**A warning to know about:** the rule of thumb says an IV above 0.5 is "suspicious" (possible leakage). Here it
applies to utilization and past-due history. These are genuine behavioral risk signals, and they describe the borrower's past, so it is
plausible for them to be that strong. But from the file alone we cannot prove they were all measured *before* the 2-year window, and in a real project
you would check that with the data owner.

```python
# Variable selection rule: keep variables with IV >= 0.02
selected_vars = iv_summary.index[iv_summary["IV"] >= 0.02].tolist()
dropped_vars = [v for v in iv_summary.index if v not in selected_vars]

print("Selected variables:", len(selected_vars))
for v in selected_vars:
    print("  -", v)
print("Dropped (IV < 0.02):", dropped_vars)
```

## 5. Replace raw values by their WoE

**What it does:** for each selected variable, every borrower's raw value is replaced by the WoE of the bin it falls in.
The test set uses the **WoE values learned on the train set**.

**Why a bank does it:** this is the "feature engineering" step that makes the final model a simple sum of
interpretable contributions. We also check that the WoE variables are not too correlated with each other,
because strongly correlated inputs make logistic regression coefficients unstable and hard to explain.

```python
# For each variable: a dictionary {bin label -> WoE}
woe_maps = {var: woe_tables[var]["woe"].to_dict() for var in selected_vars}


def to_woe(X):
    # Convert a raw-feature DataFrame into a WoE DataFrame
    out = pd.DataFrame(index=X.index)
    for var in selected_vars:
        bins = assign_bins(X[var], bin_edges[var])
        out[var] = bins.map(woe_maps[var])
    return out


X_train_woe = to_woe(X_train)
X_test_woe = to_woe(X_test)

# Safety check: every value must have found a WoE (no NaN)
print("NaN in train WoE:", X_train_woe.isna().sum().sum())
print("NaN in test WoE :", X_test_woe.isna().sum().sum())
X_train_woe.head()
```

```python
# Correlation between the WoE variables
corr = X_train_woe.corr().round(2)
display(corr)

# The 3 most correlated pairs
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
print("Most correlated pairs:")
print(upper.stack().sort_values(ascending=False).head(3))
```

**Interpretation:** the correlations between WoE variables are low to moderate (the largest pairs are printed above), so there is
no serious multicollinearity and we keep all variables. Moderate correlation is expected among the three past-due variables (they measure similar behavior).

## 6. Logistic regression on the WoE variables

**What it does:** fits a logistic regression that predicts the probability of default:

`ln( PD / (1 − PD) ) = b0 + b1·WoE1 + b2·WoE2 + ...`

**Why a bank does it:** logistic regression is the industry standard for PD scorecards because it is
transparent (each variable has one coefficient), stable, easy to monitor, and accepted by regulators.
Because every input is a WoE, we expect **all coefficients to be negative**: a *higher* WoE means *safer*, so it must *lower* the risk.
A positive coefficient would be a red flag (usually caused by correlated variables).

We use **no class weighting**: re-balancing the classes would distort the predicted probabilities, and we want
*honest* PDs (important for IFRS 9 / IRB where the PD value itself is used, not only the ranking).
`C=np.inf` means "no regularization", so we get the plain textbook logistic regression.

```python
model = LogisticRegression(C=np.inf, max_iter=1000)
model.fit(X_train_woe, y_train)

coefs = pd.Series(model.coef_[0], index=selected_vars, name="coefficient")
print("Intercept (b0):", round(model.intercept_[0], 4))
print()
print(coefs.round(4).to_frame())
print()
print("All coefficients negative (expected sign)?", bool((coefs < 0).all()))
```

**Interpretation:**

- All ten coefficients are **negative**, as expected: a safer bin (higher WoE) always lowers the predicted risk.
- Utilization (−0.61), the past-due counts (−0.37 to −0.53) and age (−0.37) carry a lot of weight.
- `DebtRatio` has the largest coefficient (−0.90) even though its IV is low (0.07). Most likely this comes from the overlap between `DebtRatio` and
  `MonthlyIncome`: they are correlated and their "Missing" bins flag almost the same borrowers, so the model shares the signal between them in a slightly odd way.
  Lesson: with correlated variables, do not over-interpret a single coefficient; judge the model by its overall performance and by the signs.
- `NumberOfOpenCreditLinesAndLoans` (−0.10) and `MonthlyIncome` (−0.07) have small coefficients: they hardly move the score once the other variables are in the model.

## 7. Convert the model into a points-based scorecard

**What it does:** turns the log-odds of the model into **points**, so that a human can read a score instead of a probability.
We fix three parameters:

- **Base score = 600** at **odds = 50:1** (50 goods for 1 bad, i.e. PD = 1/51 ≈ 1.96%)
- **PDO = 20**: every +20 points **double** the good:bad odds

The standard formulas:

- `Factor = PDO / ln(2)`
- `Offset = BaseScore − Factor × ln(BaseOdds)`
- `Score = Offset + Factor × ln(odds_good)` where `odds_good = (1 − PD) / PD = −(b0 + Σ bi·WoEi)`

So the score is a sum of one block of points per variable:

`points(variable i, bin j) = −bi × WoE_ij × Factor + (Offset − b0 × Factor) / n`   (n = number of variables)

**Why a bank does it:** a scorecard is easy to explain to a loan officer, a customer or a regulator
("you lost 25 points because your credit utilization is above 90%"), it can be computed on paper or in any system
without a model library, and cut-offs/risk grades are simple score ranges. We round each points value to an integer, like real scorecards do.

```python
BASE_SCORE = 600
BASE_ODDS = 50       # 50 goods for 1 bad
PDO = 20

factor = PDO / np.log(2)
offset = BASE_SCORE - factor * np.log(BASE_ODDS)
n_vars = len(selected_vars)
b0 = model.intercept_[0]

print(f"Factor = {factor:.4f}")
print(f"Offset = {offset:.4f}")
```

```python
# Build the points table: one row per (variable, bin)
rows = []
points_maps = {}   # {variable: {bin label: points}}

for var in selected_vars:
    table = woe_tables[var]
    points_maps[var] = {}
    for bin_label, row in table.iterrows():
        pts = -coefs[var] * row["woe"] * factor + (offset - b0 * factor) / n_vars
        pts = int(round(pts))
        points_maps[var][bin_label] = pts
        rows.append({
            "variable": var,
            "bin": bin_label,
            "n_borrowers": int(row["total"]),
            "default_rate": round(row["default_rate"], 4),
            "woe": round(row["woe"], 4),
            "points": pts,
        })

points_table = pd.DataFrame(rows)
points_table
```

```python
# A function that gives the score of any borrower = sum of points of his bins
def compute_score(X):
    total = pd.Series(0, index=X.index)
    for var in selected_vars:
        bins = assign_bins(X[var], bin_edges[var])
        total = total + bins.map(points_maps[var])
    return total


# Score -> PD (inverse of the score formula)
def score_to_pd(score):
    odds_good = np.exp((score - offset) / factor)
    return 1 / (1 + odds_good)


score_train = compute_score(X_train)
score_test = compute_score(X_test)

print("Score range on train:", score_train.min(), "to", score_train.max())
print("Score at the base odds should give PD of about 1/51 =", round(1 / 51, 4))
print("Check: PD at score 600 =", round(float(score_to_pd(600)), 4))
```

```python
# Sanity check: the rounded-points scorecard should reproduce the logistic regression PD almost exactly
pd_model_test = model.predict_proba(X_test_woe)[:, 1]
pd_score_test = score_to_pd(score_test)

print("Mean |PD model - PD scorecard| on test:", round(float(np.mean(np.abs(pd_model_test - pd_score_test))), 5))
print("AUC from model probabilities :", round(roc_auc_score(y_test, pd_model_test), 4))
print("AUC from scorecard points    :", round(roc_auc_score(y_test, -score_test), 4))
```

**How to read the points table:** more points = safer. The score is the sum of 10 blocks of points, one per variable.

- The score can be explained variable by variable. For example, going from utilization ≤ 5% (81 points) to utilization above 100% (19 points) costs **62 points**,
  which is more than 3 × PDO, i.e. the good:bad odds are divided by about 8 to 9.
- The widest point ranges are for utilization (19 to 81), 90+ days late (9 to 62), 30-59 days late (18 to 64), age (50 to 68) and debt ratio (40 to 62).
  Income, open credit lines and dependents only move the score by a few points (consistent with their small coefficients).
- On the test set the score ranges from 404 to 627 with a mean around 582. The reference point works as designed: a score of 600 corresponds to a PD of 1.96% (odds of 50:1).
- Rounding the points to integers costs almost nothing: the scorecard PD differs from the model PD by about 0.001 on average and the AUC is the same to 3 decimals.

## 8. Evaluation on the test set

**What it does:** measures how good the scorecard is on borrowers it has never seen, with:

- **AUC**: the probability that a random bad gets a *lower* score than a random good. 0.5 = coin flip, 1 = perfect.
- **Gini** = 2 × AUC − 1 (the same information rescaled from 0 to 1; the number credit risk people quote).
- **KS** (Kolmogorov–Smirnov): the maximum gap between the cumulative share of bads and goods along the score. Higher = better separation.
- **Score distribution plot**: do goods and bads sit at different score levels?
- **Calibration check**: when the model says "PD = 10%", do about 10% of those borrowers really default?

**Why a bank does it:** two different qualities are checked. **Discrimination** (AUC, Gini, KS) = can the model
*rank* risky borrowers above safe ones (needed for accept/reject decisions and risk grades). **Calibration** = are the PD *levels*
right (needed for IFRS 9 provisions and IRB capital, where the PD number itself enters the formulas).
A model can rank very well and still be badly calibrated, so both are needed.
We also show the train numbers: a big gap between train and test would signal overfitting.

```python
def ks_statistic(y_true, risk_score):
    # KS = largest vertical gap between the ROC curve and the diagonal
    fpr, tpr, _ = roc_curve(y_true, risk_score)
    return float(np.max(tpr - fpr))


results = {}
for name, y_true, score in [("train", y_train, score_train), ("test", y_test, score_test)]:
    risk = -score    # a LOW score means HIGH risk, so we flip the sign for the metrics
    auc = roc_auc_score(y_true, risk)
    results[name] = {"AUC": auc, "Gini": 2 * auc - 1, "KS": ks_statistic(y_true, risk)}

metrics = pd.DataFrame(results).T
metrics.round(4)
```

```python
# Score distribution (test set) + ROC curve
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

bins = np.arange(score_test.min() - 5, score_test.max() + 10, 10)
axes[0].hist(score_test[y_test == 0], bins=bins, alpha=0.6, density=True, label="Good (no default)", color="seagreen")
axes[0].hist(score_test[y_test == 1], bins=bins, alpha=0.6, density=True, label="Bad (default)", color="firebrick")
axes[0].set_xlabel("Score")
axes[0].set_ylabel("Density")
axes[0].set_title("Score distribution, test set")
axes[0].legend()

fpr, tpr, _ = roc_curve(y_test, -score_test)
axes[1].plot(fpr, tpr, label=f"Scorecard (AUC = {results['test']['AUC']:.3f})")
axes[1].plot([0, 1], [0, 1], "--", color="grey", label="Random")
axes[1].set_xlabel("False positive rate")
axes[1].set_ylabel("True positive rate")
axes[1].set_title("ROC curve, test set")
axes[1].legend()

plt.tight_layout()
plt.savefig("figures/score_distribution_and_roc.png", dpi=120)
plt.show()
```

```python
# Calibration check: split the test set into 10 groups of equal size (deciles of predicted PD),
# then compare the average predicted PD with the observed default rate in each group.
cal = pd.DataFrame({"pd_pred": score_to_pd(score_test), "bad": y_test.values})
cal["decile"] = pd.qcut(cal["pd_pred"].rank(method="first"), 10, labels=False) + 1   # 1 = safest, 10 = riskiest

cal_table = cal.groupby("decile").agg(
    n=("bad", "count"),
    avg_pd_predicted=("pd_pred", "mean"),
    observed_default_rate=("bad", "mean"),
)
print(cal_table.round(4))
print()
print(f"Overall: predicted PD = {cal['pd_pred'].mean():.4f} | observed default rate = {cal['bad'].mean():.4f}")

# Plot
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.plot(cal_table["avg_pd_predicted"], cal_table["observed_default_rate"], "o-", label="Deciles")
top = max(cal_table["avg_pd_predicted"].max(), cal_table["observed_default_rate"].max()) * 1.05
ax.plot([0, top], [0, top], "--", color="grey", label="Perfect calibration")
ax.set_xlabel("Average predicted PD")
ax.set_ylabel("Observed default rate")
ax.set_title("Calibration, test set")
ax.legend()
plt.tight_layout()
plt.savefig("figures/calibration.png", dpi=120)
plt.show()
```

```python
# Is the gap in the safest deciles caused by rounding the points? Compare with the unrounded model, on test AND train.
for name, X_woe, y_true in [("test", X_test_woe, y_test), ("train", X_train_woe, y_train)]:
    chk = pd.DataFrame({"pd_model": model.predict_proba(X_woe)[:, 1], "bad": y_true.values})
    chk["decile"] = pd.qcut(chk["pd_model"].rank(method="first"), 10, labels=False) + 1
    print(f"\nUnrounded model, {name} set:")
    print(chk.groupby("decile").agg(avg_pd_model=("pd_model", "mean"), observed=("bad", "mean")).round(4).T.to_string())
```

**Interpretation (real numbers from this run):**

- **Discrimination:** on the test set, AUC = 0.857, Gini = 0.715, KS = 0.562. Train and test values are almost identical (AUC 0.8575 vs 0.8574), so there is **no sign of overfitting**.
- **Score distribution:** the bads (red) are concentrated at lower scores than the goods (green), with overlap in the middle. That overlap is normal in credit scoring.
- **Calibration:** on average the predicted PD (6.68%) equals the observed default rate (6.68%). In the riskiest decile the model predicts 35.0% against 36.2% observed.
  The gap is larger in relative terms at the safe end: it **over-predicts** the safest deciles (decile 1: 1.0% predicted vs 0.5% observed) and slightly **under-predicts** decile 9 (10.2% vs 11.9%).
- The check above shows the same pattern for the unrounded model, on both train and test, so it is **not** caused by rounding the points or by overfitting.
  It comes from the model form (a simple sum of WoE terms). A bank would either accept it, or fix it with a recalibration step, more bins or interaction terms.
- Reminder: good discrimination (Gini 0.71) does not prove calibration. We checked both, and calibration is acceptable overall but imperfect at the extremes.

## 10. Limits of this project (be honest about them in interviews)

- It is an **application-style scorecard on a public dataset**, not a production IFRS 9 or IRB model.
- The target is "serious delinquency (90+ days) within 2 years", **not** a 12-month regulatory default definition.
- **IFRS 9** also needs lifetime PDs, forward-looking macro scenarios, staging (SICR), LGD and EAD. **IRB** needs PD calibrated to a long-run
  average default rate, a downturn LGD, rating grades, and independent validation. None of that is done here.
- No out-of-time validation (the data has no date), no monotonic-binning algorithm (bins are chosen by hand from default-rate tables),
  no stability monitoring (PSI), no reject inference.

Possible next steps: automatic optimal binning, PSI between train and test, a cut-off / risk-grade analysis, a comparison with gradient boosting.
